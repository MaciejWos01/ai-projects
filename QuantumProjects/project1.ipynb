```python
from qiskit import QuantumRegister, ClassicalRegister, QuantumCircuit, execute, Aer, IBMQ
from qiskit.compiler import transpile, assemble
from qiskit.tools.jupyter import *
from qiskit.visualization import *
from numpy import pi
backend = Aer.get_backend('qasm_simulator')
```

### Task 1
Z-type projection measurement – reading of the qubit state

```python
n = 4
seria = 2048
nx = n 
qx = QuantumRegister(nx) 
cx = ClassicalRegister(nx) 
circuitX = QuantumCircuit(qx, cx)
circuitX.measure(qx[0], cx[0])
```

```python
circuitX.draw(output='mpl')
```

```python
results = [[], []]
num_executions = 3

for _ in range(num_executions):
    job_simX = execute(circuitX, backend, shots=seria)
    sim_resultX = job_simX.result()
    counts = sim_resultX.get_counts(circuitX)
    results[0].append(counts)
    probabilities = {state: count / seria for state, count in counts.items()}
    results[1].append(probabilities)
```

```python
plot_histogram(results[0], legend=[f"Execution {i+1}" for i in range(num_executions)], title='Distribution [counts]')
```

```python
plot_histogram(results[1], legend=[f"Execution {i+1}" for i in range(num_executions)], title='Probability')
```

### Task 5

```python
from qiskit import BasicAer
from qiskit.visualization import plot_state_city, plot_bloch_multivector
from qiskit.visualization import plot_state_paulivec, plot_state_hinton
from qiskit.visualization import plot_state_qsphere

n = 4
seria = 2048
nx = n 
qx = QuantumRegister(nx) 
cx = ClassicalRegister(nx) 
circuit = QuantumCircuit(qx, cx)
circuit.x(qx[0])
circuit.measure(qx[0], cx[0])

backend = BasicAer.get_backend('statevector_simulator')
result = backend.run(transpile(circuit, backend)).result()
psi = result.get_statevector(circuit)
circuit.draw(output='mpl')
```

```python
plot_state_city(psi)
```

```python
plot_state_hinton(psi)
```

```python
plot_state_qsphere(psi)
```

```python
plot_bloch_multivector(psi)
```
